## What is Vectorization?

Vectorization is the process of applying operations to entire arrays (vectors, matrices) at once, rather than element by element using explicit loops.

**Why is it faster?**

1.  **Underlying C Implementation:** Libraries like NumPy perform vectorized operations using highly optimized, pre-compiled C code. This avoids the overhead of interpreting Python code for each element in a loop.
2.  **Reduced Python Overhead:** Python loops have overhead associated with type checking, function call lookups, and object management for each iteration. Vectorization minimizes this.
3.  **CPU Optimizations:** Optimized C code can often take advantage of modern CPU features like SIMD (Single Instruction, Multiple Data) instructions, which perform the same operation on multiple data points simultaneously.

```python
# Define the size of the data
# You can change this value to see how performance scales
#data_size = 1_000_000 # 1 million elements
data_size = 10_000_000 # 10 million elements (might take longer for the loop)

print(f"Using data size: {data_size:,}")
```

### 1. Loop-Based Addition

```python
# Generate sample data (Python lists)
list_a = list(range(data_size))
list_b = list(range(data_size))

print(f"Created two lists of size {len(list_a):,}")

# Time the loop-based approach
start_time_loop = time.time()
result_loop = loop_addition(list_a, list_b)
end_time_loop = time.time()
loop_duration = end_time_loop - start_time_loop

print(f"Loop-based addition took: {loop_duration:.6f} seconds")
```

### 2. Vectorized Addition (NumPy)

```python
# Generate sample data (NumPy arrays)
array_a = np.arange(data_size, dtype=np.int64) # Using int64 for consistency
array_b = np.arange(data_size, dtype=np.int64)

print(f"Created two NumPy arrays of shape {array_a.shape}")

# Time the vectorized approach
start_time_vec = time.time()
result_vec = vectorized_addition(array_a, array_b)
end_time_vec = time.time()
vec_duration = end_time_vec - start_time_vec

print(f"Vectorized addition took: {vec_duration:.6f} seconds")
```

### 3. Performance Comparison

```python
# Calculate and print the speedup
if vec_duration > 0 and loop_duration > 0:
    speedup = loop_duration / vec_duration
    print(f"Vectorization Speedup: {speedup:.2f}x")
    print(f"(The vectorized version was {speedup:.2f} times faster than the loop-based version.)")
elif loop_duration == 0:
    print("Loop duration was too small to measure accurately.")
else: # vec_duration is 0 or very small
    print("Vectorized operation was extremely fast!")
```

## Conclusion & Takeaway

As you can see, the vectorized approach using NumPy is significantly faster than the equivalent operation implemented with a standard Python loop. The difference becomes even more pronounced as the data size increases.

**Key Takeaway:** Whenever performing numerical operations on sequences of data (lists, arrays), **always** prefer vectorized implementations using libraries like NumPy or Pandas over explicit Python loops for substantial performance gains.
